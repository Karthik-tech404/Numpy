```python
🔵 What is NumPy?

NumPy (Numerical Python) is a Python library designed for working with numerical data and performing mathematical operations efficiently.

Think of it as a powerful tool for handling arrays, matrices, and large amounts of numerical data.

✅ Advantages of NumPy
Fast
NumPy performs numerical calculations much faster than normal Python lists.
It is especially useful when working with large datasets.
Efficient memory usage
NumPy arrays generally use memory more efficiently than Python lists.
Easy mathematical operations
You can perform mathematical calculations on entire arrays efficiently.
Supports multidimensional data
It can work with 1D, 2D, 3D, and higher-dimensional arrays.
Many built-in functions
Provides functions for statistics, mathematics, linear algebra, random numbers, etc.
Foundation for Data Science
Many Python libraries used in Data Science and Machine Learning work with NumPy or are built around similar array concepts.
🎯 Benefits of NumPy

The main benefits are:

⚡ Faster numerical calculations
💾 Better memory efficiency
📊 Easy handling of numerical datasets
🧮 Powerful mathematical operations
📐 Matrix and linear algebra support
🎲 Random number generation
🤖 Useful for Machine Learning
📈 Useful for Data Science
🔗 Works well with other Python libraries
🔧 Uses of NumPy

NumPy is commonly used for:

1. Data Science

Working with and processing numerical datasets.

2. Machine Learning

Preparing numerical data and performing mathematical operations required by ML algorithms.

3. Artificial Intelligence

Many AI operations involve large amounts of numerical data.

4. Scientific Computing

Used for scientific and engineering calculations.

5. Statistics

Calculating things such as mean, median, standard deviation, etc.

6. Linear Algebra

Working with vectors, matrices, matrix multiplication, and related operations.

7. Image Processing

Images can be represented as numerical arrays, so NumPy can be used to manipulate image data.

8. Simulations

Useful for generating and processing numerical data for simulations.

❌ Disadvantages of NumPy

NumPy is powerful, but it has some limitations:

Mainly designed for numerical data
It isn't as flexible as normal Python lists for storing completely different types of objects.
Requires learning
Arrays, dimensions, shapes, indexing, broadcasting, etc., can initially be confusing.
Not ideal for every type of data
For tabular data with columns of different types, Pandas is usually more convenient.
Memory considerations
Although NumPy is memory-efficient compared with Python lists in many cases, very large arrays can still consume significant memory.
Less suitable for general-purpose programming
NumPy is specifically focused on numerical/scientific computing rather than replacing Python's normal data structures.


```

```python
#Broadcasting & Vectorisation
#Vectorisation means performing an operation on an entire NumPy array at once instead of manually processing each element one by one.
import numpy as np
arr = np.array([1,2,3,4,5,6])
sum = arr * 3
print(sum)

'''
🧠 Why is Vectorisation useful?
Without vectorisation

You would typically need a loop to process every value.

With vectorisation

You perform the operation directly on the array.

Benefits:

⚡ Faster calculations
🧹 Cleaner code
📊 Good for large numerical datasets
🤖 Very useful in Machine Learning
🔢 Efficient mathematical operations


'''

#Broadcasting allows NumPy to perform operations between arrays with different shapes when their dimensions are compatible.
arr = np.array([10, 20, 30, 40])

arr_1 = np.array([[10,20,30,40],[20,30,40,50]])

result = arr + arr_1

print(result)

'''
⭐ Broadcasting Rules

This is the part you should understand carefully.

NumPy compares dimensions from right to left.

Two dimensions are compatible when:

Rule 1

They are the same size.

Example:

3
3

Compatible ✅

Rule 2

One of the dimensions is 1.

Example:

3
1

Compatible ✅

Rule 3

A missing dimension can be treated as 1.

❌ When Broadcasting Doesn't Work

Consider:

Array A → (2, 3)

Array B → (2,)

Let's compare from the right:

A:  2  3
B:     2

Compare:

3 vs 2

They're different and neither is 1.

So they aren't compatible ❌.

This produces a broadcasting error.

🟢 Practical Example: Adding a Constant

Suppose you have student marks:

[50, 60, 70, 80]

You want to give everyone 5 bonus marks.

Broadcasting allows you to add 5 to the entire array:

marks = np.array([50, 60, 70, 80])

new_marks = marks + 5

print(new_marks)

Result:

[55 65 75 85]
🟢 Vectorisation vs Broadcasting

This is important because they're related but not the same thing.

Vectorisation

Performing operations on an entire array efficiently.

Example:

array × 2

Broadcasting

Automatically adjusting compatible arrays/shapes so operations can happen between them.

Example:

2D array + 1D array

🧠 Easy memory trick

Vectorisation = HOW the operation is performed

Broadcasting = HOW different shapes work together

'''
```

```python
#Handling Missing Values in NumPy
'''
Missing values are extremely common when working with real-world data.

For example, imagine a dataset of student marks:

Student     Marks
A           85
B           72
C           ?
D           91

The ? means the value is missing.

In NumPy, the most common representation for a missing numerical value is NaN.
'''
'''
1. What is NaN?

NaN = Not a Number

It represents a value that is missing or undefined.

For example:

arr = np.array([10, 20, np.nan, 40, 50])

print(arr)

You'll see:

[10. 20. nan 40. 50.]

Notice something interesting:

The integers became floating-point values.

That's because NaN is generally associated with floating-point numerical data.

2. Checking for Missing Values

NumPy provides:

np.isnan(arr)

This checks every element.

Conceptually:

10  → False
20  → False
NaN → True
40  → False
50  → False

So you'll get a Boolean array.

🧠 Remember

np.isnan() → checks whether a value is NaN

3. Finding How Many Missing Values

You can combine isnan() with sum():

np.isnan(arr).sum()

If your array contains one NaN, the result is:

1

Why?

Because:

True  = 1
False = 0

So counting True values gives the number of missing values.

4. Finding Non-Missing Values

You can use:

~np.isnan(arr)

The ~ means NOT.

So:

np.isnan(arr)

False False True False False

After ~:

True True False True True

This can be used to select values that aren't missing.

5. Removing NaN Values

NumPy provides:

arr[~np.isnan(arr)]

This selects only the values that are not NaN.

For:

[10, 20, NaN, 40, 50]

you get:

[10, 20, 40, 50]
'''
```
